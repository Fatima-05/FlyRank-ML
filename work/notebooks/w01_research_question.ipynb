## 2. The question: decision, action, cost of a wrong call

```python
# The Question:
# Which pages should be reviewed first for a content refresh so that a small team can recover the most declining traffic with limited effort?

# Decision it improves: Prioritization of the content refresh queue.

# Who acts: A content strategist or SEO editor who has limited hours each week.

# Action they take: Open the top-ranked pages, decide whether a refresh is warranted, and either update the page or skip it.

# Cost of a wrong recommendation:
# False positive (flagging a healthy page) which leads to wasted editor time.
# False negative (missing a declining page) which leads to continued traffic loss that could have been recovered.

# Unit of analysis: One content page (content_hash_id/page-level row).

# Desired output: A ranked list of pages with a score, reason codes, and a short “why this page” explanation.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys
import pandas as pd
import json

while not os.path.exists("data/raw/content_refresh_anonymized.csv") and os.getcwd() != "/":
    os.chdir("..")

print("Working directory:", os.getcwd())
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("Rows:", len(df))
print("Declining rate:", round((df["trend_direction"].str.lower() == "down").mean(), 3))

if os.path.exists("outputs/model_results.json"):
    res = json.load(open("outputs/model_results.json"))
    base = res["baseline"]["baseline_precision_at_50"]
    rf   = res["models"]["random_forest"]["precision_at_50"]
    print(f"Hand-rule Precision@50: {base:.3f}")
    print(f"Random forest Precision@50: {rf:.3f}")
    print(f"Lift: ~{rf/base:.1f}x")
```
